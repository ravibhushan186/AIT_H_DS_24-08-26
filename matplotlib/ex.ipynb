```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
# Draw a line in a diagram from position (0,0) to position (6,250):
xAxis = np.array([0,6])
yAxis = np.array([0,250])

plt.plot(xAxis, yAxis)
plt.show()
```

```python
years = np.array([2022, 2023, 2024, 2025, 2026])
python = np.array([20, 30,25, 40, 50])

plt.plot(years, python, marker = 'o')
plt.show()
```

```python
years = np.array([2022, 2023, 2024, 2025, 2026])
python = np.array([20, 30,25, 40, 50])
java = np.array([10, 20, 30, 40, 50])

line = dict(marker ='o', markersize = 10, markerfacecolor = 'red', markeredgecolor = 'blue', markeredgewidth = 2, linestyle = '--', linewidth = 3)

plt.plot(years, python, color='green', **line)
plt.plot(years, java, color='blue', **line)
plt.title('Programming Language Popularity')
plt.xlabel('Years')
plt.ylabel('Popularity')
plt.xticks(years)
plt.ylabel('Popularity')
plt.grid(axis='both', color='gray', linestyle='--', linewidth=0.5)
plt.show()
```

```python
products = ['Laptop', 'Desktop', 'Tablet', 'Smartphone', 'Smartwatch']
sales = np.array([100, 150, 80, 200, 50])

plt.pie(sales, labels=products, autopct='%1.1f%%', explode=(0.1, 0, 0, 0, 0), shadow=True, startangle=90)
plt.title('Product Sales Distribution')
plt.show()
```

```python
products = ['Laptop', 'Desktop', 'Tablet', 'Smartphone', 'Smartwatch']
sales = np.array([100, 150, 80, 200, 50])
plt.bar(products, sales, color=['blue', 'green', 'orange', 'red', 'purple'])
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
fig, ax = plt.subplots(2,2)
ax[0,0].plot(years, python, color='green', marker='o')
ax[0,0].set_title('Python Popularity')

plt.show()
```
