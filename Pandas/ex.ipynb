```python
l = [100,200,300,400,500]
series = pd.Series(l)
series
#Note: It is automatically assigned an index 0 to n-1, where n is the number of elements in the series.
```

```python
# If you want to assign your own index, you can do so by passing the index parameter while creating the series.
l = [100,200,300,400,500]
series = pd.Series(l, index=['a','b','c','d','e'])
series
```

```python
# You can access the elements of the series using the index labels. For example, to access the element with index 'a', you can use:
series.loc['a']
```

```python
# You can also access the elements using the integer index position. For example, to access the first element of the series, you can use:
series.iloc[0]
```

```python
# Create a pandas Series from a dictionary
walk = {"Day1": 1000, "Day2": 2000, "Day3": 3000}
walk_series = pd.Series(walk)
walk_series
```

```python
walk_series.loc['Day1']+=200
walk_series
```

```python
# Create a DataFrame with custom index
employee = {"Name": ["John", "Ravi", "Bhushan"],
            "age": [25, 30, 35],
            "department": ["HR", "IT", "Finance"]}
a=pd.DataFrame(employee)
a
```

```python
# Create a DataFrame with custom index
employee = {"Name": ["John", "Ravi", "Bhushan"],
            "age": [25, 30, 35],
            "department": ["HR", "IT", "Finance"]}
a=pd.DataFrame(employee,["emp1","emp2","emp3"])
a
```

```python
# Select the row for employee with index 'emp1'
a.loc['emp1']
```

```python
# Add a new column for Salary
a["Salary"] = [50000, 60000, 70000]
a
```

```python
# Add a new row
new_row=pd.DataFrame({"Name":["Ramesh"],"age":[28],"department":["IT"],"Salary":[65000]},index=["emp4"])
new_row
```

```python
# Concatenate the new row to the existing DataFrame
a = pd.concat([a, new_row])
a
```
